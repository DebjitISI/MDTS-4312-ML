```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split,KFold,GridSearchCV
from sklearn.svm import SVC
import ISLP.data
```

```python
data = load_breast_cancer()
```

```python
data
```

```python
X = data.data
y = data.target
```

```python
st = StandardScaler()
X_scaled = st.fit_transform(X)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X_scaled,y,test_size=0.25)
```

```python
sv = SVC(kernel='linear')
model = sv.fit(X_train,y_train)
print(f"Train Acc:{model.score(X_train,y_train)}")
print(f"Test Acc:{model.score(X_test,y_test)}")
```

```python
sv = SVC(kernel='linear',C=10)
model = sv.fit(X_train,y_train)
print(f"Train Acc:{model.score(X_train,y_train)}")
print(f"Test Acc:{model.score(X_test,y_test)}")
```

```python
c = range(11)
sv = SVC(kernel='linear')
prams = {'C':[0.2,0.5,1,2,3,4,5,9,10]}
kf = KFold(n_splits=5,shuffle=True,random_state=0)

gd = GridSearchCV(sv,prams,cv=kf,scoring='neg_mean_squared_log_error')
```

```python
gd.fit(X_train,y_train)
```

```python
gd.best_params_
```

```python
sv = SVC(kernel='linear',C=0.5)
model = sv.fit(X_train,y_train)
print(f"Train Acc:{model.score(X_train,y_train)}")
print(f"Test Acc:{model.score(X_test,y_test)}")
```

```python
from ISLP import load_data
carseats = load_data('Carseats')
carseats.head()
```

```python
data = carseats.drop(columns=['ShelveLoc','Urban','US'])
```

```python
data['Sales_cat'] = np.where(data['Sales'] > 8, 1, 0)
```

```python
data
```

```python
X = data.drop(columns=['Sales_cat'])
y = data['Sales_cat']
```

```python
st = StandardScaler()
X_scaled = st.fit_transform(X)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X_scaled,y,test_size=0.25,random_state=0)
```

```python
c = range(11)
sv = SVC(kernel='linear')
prams = {'C':[0.2,0.5,1,2,3,4,5,9,10]}
kf = KFold(n_splits=5,shuffle=True,random_state=0)

gd = GridSearchCV(sv,prams,cv=kf,scoring='neg_mean_squared_log_error',n_jobs=-1)
```

```python
gd.fit(X_train,y_train)
```

```python
gd.best_params_
```

```python
sv = SVC(kernel='linear',C=1)
model = sv.fit(X_train,y_train)
print(f"Train Acc:{model.score(X_train,y_train)}")
print(f"Test Acc:{model.score(X_test,y_test)}")
```
